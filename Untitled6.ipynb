```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

A RunnableLambda in LangChain is a wrapper that turns a standard custom function into a pipe-compatible component

```python
from langchain_core.runnables import RunnableLambda
```

```python
def clean_text(text: str) -> str:
    return text.strip().lower()

def add_prefix(text: str) -> str:
    return f"Processed: {text}"
```

```python
# chains the steps, so the output of clean_text becomes the input to add_prefix.
chain = RunnableLambda(clean_text) | RunnableLambda(add_prefix)

result = chain.invoke("  !HELLO WORLD!  ")
print(result)
```

```python
# Component 1: add 5 to the input number
add_five = RunnableLambda(lambda x: {"value": x["number"] + 5})
```

```python
# Component 2: multiply the previous output by 2
multiply_by_two = RunnableLambda(lambda x: x["value"] * 2)
```

```python
# Native LangChain pipeline
arithmetic_chain = add_five | multiply_by_two
```

```python
result = arithmetic_chain.invoke({"number": 10})
print(result)
```

RAG
Retrieval-Augmented Generation (RAG) is a framework that connects a Large Language Model (LLM) to external data sources—like your own documents or private databases—at query time to provide accurate, up-to-date, and grounded answers.

```python
import os
from langchain_community.document_loaders import PyPDFLoader, TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
```

```python
def load_documents():
    docs = []
    if os.path.exists("Sample.txt"):
        docs.extend(TextLoader("Sample.txt").load())
    if not docs:
        raise FileNotFoundError("Add sample.pdf or sample.txt to the folder.")
    return docs
```

```python
def split_documents(documents):
    splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100)
    return splitter.split_documents(documents)
```

```python
def create_vectorstore(chunks):
    embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
    return FAISS.from_documents(chunks, embeddings)
```

```python
def ask_rag(question, vectorstore):
    retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
    docs = retriever.invoke(question)
    context = "\n\n".join(doc.page_content for doc in docs)
    prompt = f"""
    Answer the question using only the context below.
    If the answer is not in the context, say you do not know.

    Context:
    {context}

    Question:
    {question}
    """
    llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
    response = llm.invoke(prompt)
    return response.content, docs
```

```python
def main():
    documents = load_documents()
    chunks = split_documents(documents)
    vectorstore = create_vectorstore(chunks)

    #question = "What is this document mainly about?"
    #question = "What is RAG?"
    #question = "What are the five steps in a RAG workflow?"
    #question = "Why is RAG useful?"
    #question = "What is machine learning?"
    question = "How far the sun is from earth?"
    answer, source_docs = ask_rag(question, vectorstore)

    print("QUESTION:")
    print(question)
    print("\nANSWER:")
    print(answer)
    print("\nSOURCES:")
    for i, doc in enumerate(source_docs, start=1):
        print(i, doc.metadata)
```

```python
main()
```
